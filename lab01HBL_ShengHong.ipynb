## Task 1: Grayscale Loading

1. Load the image `images2/komodo.jpg` in **Grayscale** mode.
2. Display it using `matplotlib.pyplot` with the correct colormap.
3. Set the title to "Grayscale Image".

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os
import glob

```

```python
img_gray = cv2.imread('images2/komodo.jpg', cv2.IMREAD_GRAYSCALE)
```

```python
print("Grayscale Image Shape", img_gray.shape)
```

```python
plt.figure(figsize=(6,6))
plt.imshow(img_gray, cmap='gray')
plt.title("Grayscale Image")
plt.axis('off')
plt.show()
```

## Task 2: Color Conversion

1. Load the image `images2/komodo.jpg` in Color (Standard).
2. Convert the image from **BGR** (OpenCV default) to **RGB**.
3. Display the result with the title "RGB Image".

```python
img_bgr = cv2.imread('images2/komodo.jpg')
```

```python
print("colour Image Shape:", img_bgr.shape)
```

```python
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6,6))
plt.imshow(img_rgb)
plt.title("RGB Image")
plt.axis('off')
plt.show()
```
