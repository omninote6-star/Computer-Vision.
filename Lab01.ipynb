```python
img_rgba = cv2.imread('images/stickanimalsRGBA.png', cv2.IMREAD_UNCHANGED)
```

```python
print("RGBA Image Shape:", img_rgba.shape)

if img_rgba.shape[2] == 4:
    print("Image has 4 channels: R, G, B, Alpha")
```

```python
img_rgba_corrected = cv2.cvtColor(img_rgba, cv2.COLOR_BGRA2RGBA)

plt.figure(figsize=(6,6))
plt.imshow(img_rgba_corrected)
plt.title("RGBA Image with Transparency")
plt.axis('off')
plt.show()
```

```python
folder_path = 'images2'

file_names = os.listdir(folder_path)
print(f"Found{len(file_names)} files: {file_names}")
```

```python
plt.figure(figsize=(15, 5))

for i,file_name in enumerate(file_names):
    img_path = os.path.join(folder_path, file_name)

    img = cv2.imread(img_path)

    if img is None:
        continue

    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    plt.subplot(1,len(file_names), i + 1)
    plt.imshow(img)
    plt.title(file_name)
    plt.axis('off')

    plt.tight_layout()
    plt.show()
```
