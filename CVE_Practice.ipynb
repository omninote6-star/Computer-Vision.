```python
#task 2 import numpy
import numpy as np
```

```python
# #task3: Convert X_values into a numpy array
X_values = np.array(X_values)
```

```python
#task4: Convert y_values into a numpy array
y_values = np.array(y_values)
```

```python
#task5: Apply image normalization
X_values = X_values / 255 
```

```python
#task6: Reshape X_values to have a 4d format
X_values = X_values.reshape(-1, 512, 512, 1)
```

```python
#task9: Import train test split
from sklearn.model_selection import train_test_split
```

```python
#task10: Extraction of the data as test dataset
X_OTHER, X_test, y_OTHER, y_test = train_test_split(X_values, y_values, test_size=0.1, stratify=y_values, random_state=42)
```

```python
#Task11: Split the remaining data
X_train, X_val, y_train, y_val = train_test_split(X_OTHER, y_OTHER, test_size=0.3, stratify=y_OTHER, random_state=42)
```

```python
#task12: Import Sequential and layers from tensorflow
from keras.models import Sequential
```

```python
#task13: import layers fork keras
from keras.layers import Dense, Flatten, Conv2D
```

```python
#task14: Create a model use keras sequential API
model = Sequential()
model.add(Conv2D(
    filters=64, 
    kernel_size=(3,3), 
    strides=1, padding='same',
    activation='relu', 
    input_shape=(512, 512, 1)
))
model.add(Flatten())
model.add(Dense(2, activation= 'softmax'))
model.summary()
```

```python
# task15: Compile the model
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics='accuracy')
```

```python
# task16: Train the model
history = model.fit(X_train, y_train, epochs=20, validation_data=(X_val, y_val), verbose=1)
```

```python
# task 17: visualise the training and val accuaracy
plt.plot(history.history['Loss'], label= 'Training Loss')
plt.plot(history.history['val_loss'], label= 'Validation Loss')
plt.title('model Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.show()
```

```python
#task 18 visualise the training and val accuaracy
plt.plot(history.history['accuracy'], label= 'Training Accuracy')
plt.plot(history.history['val_accuracy'], label= 'Validation Accuracy')
plt.title('model Accuracy')
plt.xlabel('Epochs')    
plt.ylabel('Accuracy')
plt.legend()
plt.show()
```

```python
# task19: predict the classes base on X_test
y_pred = model.predict(X_test).argmax(axis=1)
```

```python
# task20 Import classfication from sk learn
from sklearn.metrics import classification_report
```

```python
# task 21 Display classfication report
print(classification_report(y_test, y_pred))
```
